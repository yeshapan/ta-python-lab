## **Problem 1 (Easy)**
Given the list `numbers = [10, 20, 30, 40, 50, 60, 70]`, use basic list slicing `list[start:stop]` to extract and print the sub-list `[30, 40, 50]`.

```python
numbers = [10, 20, 30, 40, 50, 60, 70]

# Index 2 holds value 30.
# We want up to value 50 (index 4), so the stop index must be 5 because stop_index is EXCLUSIVE.
sub_list = numbers[2:5]

print("Extracted sub-list:", sub_list)
```

## **Problem 2 (Easy)**
Create a list `alphabet = ["a", "b", "c", "d", "e", "f", "g", "h"]`. Use slicing with a step value to print every second element starting from index 0.

```python
alphabet = ["a", "b", "c", "d", "e", "f", "g", "h"]

# Slicing syntax: list[start : stop : step]
# Leaving start and stop empty defaults to beginning and end.
# Step value of 2 skips every second element.
every_second = alphabet[::2]

print("Every second element:", every_second)
```

## **Problem 3 (Easy)**
Given the list `data = [100, 200, 300, 400, 500]`, use slicing with a negative step `[::-1]` to create and print a reversed copy of the list.

```python
data = [100, 200, 300, 400, 500]

# A step of -1 iterates backwards from end to start.
# This creates a brand new reversed copy without modifying the original list.
reversed_data = data[::-1]

print("Original list:", data)
print("Reversed copy:", reversed_data)
```

## **Problem 4 (Easy)**
Given `scores = [88, 42, 95, 63, 71]`, write a program that uses the built-in functions `min()`, `max()`, and `sum()` to print the lowest score, highest score, and the total sum of all scores.

```python
scores = [88, 42, 95, 63, 71]

# min() returns the smallest numerical element
lowest_score = min(scores)

# max() returns the largest numerical element
highest_score = max(scores)

# sum() calculates the running total of all items in an iterable
total_sum = sum(scores)

print("Lowest Score:", lowest_score)
print("Highest Score:", highest_score)
print("Total Sum:", total_sum)
```

## **Problem 13 (Moderate)**
Write a program that takes a user-entered word and converts it into a list of characters. Check if the list is equal to its slice reverse (`lst == lst[::-1]`) to determine if the word is a Palindrome (e.g., `"radar"`). Print `True` or `False`.

```python
user_word = input("Enter a word: ")

# Convert input word to lowercase and split into a list of characters
char_list = list(user_word.lower())

# A palindrome reads identically forward and backward
is_palindrome = (char_list == char_list[::-1])

print("Character List:", char_list)
print("Is Palindrome?", is_palindrome)
```

## **Problem 14 (Moderate)**
Given `values = [12, 0, 45, 0, 78, 0, 90]`, write a program using a `while` loop and the `.remove()` method to delete all occurrences of `0` from the list. Print the cleaned list.

```python
values = [12, 0, 45, 0, 78, 0, 90]

# .remove() only deletes the FIRST occurrence of a value it finds.
# We use a while loop to repeatedly check if 0 still exists in the list.
while 0 in values:
    values.remove(0)

print("Cleaned list without zeros:", values)
```

## **Problem 15 (Moderate)**
Create a list `items = ["pen", "notebook", "eraser"]`. Ask the user for an item name. Use an `if-else` statement paired with the `in` operator to check if it exists in the list. If it exists, print its index using `.index()`; if not, print `"Item not found"`.

## **Problem 16 (Thinking-Based / Application)**
### **The Leaderboard Top-3 Filter**
You are designing the post-match summary for an esports tournament. The game engine generates an unsorted list of player scores: `all_scores = [340, 890, 120, 950, 670, 890, 410]`.
* Use the `.sort(reverse=True)` method to organize scores from highest to lowest.
* Use list slicing to extract only the top 3 highest scores.
* Calculate the average score of these top 3 players using `sum()` and `len()`.
* Print the top 3 scores list and their calculated average score.

```python
all_scores = [340, 890, 120, 950, 670, 890, 410]

# 1. Sort the list descending in-place
all_scores.sort(reverse=True)

# 2. Extract top 3 elements via slicing (indices 0, 1, 2)
top_3_scores = all_scores[:3]

# 3. Calculate average using sum() and len()
average_top_3 = sum(top_3_scores) / len(top_3_scores)

# 4. Display results
print("Top 3 Scores:", top_3_scores)
print("Average of Top 3:", round(average_top_3, 2))
```

## **Problem 17 (Thinking-Based / Application)**
### **The E-Commerce Price Range Filter**
An online shopping platform allows users to apply a price filter. The system stores product prices in a list: `product_prices = [15, 80, 120, 45, 300, 210, 95, 50]`.
* Ask the user to input two numbers: `min_price` and `max_price`.
* Iterate through `product_prices` and create a new list `filtered_prices` containing only items that fall within `[min_price, max_price]` (inclusive).
* If no products match, print `"No items match your filter"`. Otherwise, print the sorted `filtered_prices` list using `.sort()`.

```python
product_prices = [15, 80, 120, 45, 300, 210, 95, 50]

# Accept range boundaries from user
min_price = float(input("Enter minimum price: "))
max_price = float(input("Enter maximum price: "))

filtered_prices = []

# Filter prices within bounds (inclusive)
for price in product_prices:
    if min_price <= price <= max_price:
        filtered_prices.append(price)

# Check if any items matched criteria
if not filtered_prices:
    print("No items match your filter")
else:
    filtered_prices.sort()
    print("Matching products (sorted):", filtered_prices)
```

## **Problem 18 (Thinking-Based / Application)**
### **The Playlist Re-orderer & Merger**
A user is merging two music playlists:
* `rock_playlist = ["Bohemian Rhapsody", "Hotel California"]`
* `pop_playlist = ["Shape of You", "Blinding Lights", "Bad Guy"]`
* Merge `pop_playlist` into `rock_playlist` using `.extend()`.
* Calculate the middle index using integer floor division (`len(list) // 2`).
* Use `.insert()` to place `"Stairway to Heaven"` at that exact middle index and print the updated playlist.

```python
rock_playlist = ["Bohemian Rhapsody", "Hotel California"]
pop_playlist = ["Shape of You", "Blinding Lights", "Bad Guy"]

# 1. Extend rock_playlist with pop_playlist
rock_playlist.extend(pop_playlist)

# 2. Calculate dynamic middle index using floor division
mid_index = len(rock_playlist) // 2

# 3. Insert target track at the calculated index
rock_playlist.insert(mid_index, "Stairway to Heaven")

print("Calculated Middle Index:", mid_index)
print("Updated Playlist:", rock_playlist)
```

## **Problem 19 (Thinking-Based / Application)**
### **The Duplicate Detector & Frequency Counter**
A biometric attendance system registers card swipes by ID numbers in the order they occur: `swipes = [101, 102, 101, 103, 104, 101, 102, 105]`.
* Create an empty list `unique_swipes`.
* Loop through each ID in `swipes`. If the ID is not already in `unique_swipes`, add it using `.append()`.
* After building the unique list, iterate through `unique_swipes` and use `.count()` on `swipes` to display how many times each employee swiped their card.

```python
swipes = [101, 102, 101, 103, 104, 101, 102, 105]
unique_swipes = []

# Build unique deduplicated list preserving original order
for swipe_id in swipes:
    if swipe_id not in unique_swipes:
        unique_swipes.append(swipe_id)

# Iterate through unique IDs and count occurrences in original swipes list
for swipe_id in unique_swipes:
    frequency = swipes.count(swipe_id)
    print(f"ID {swipe_id} swiped {frequency} times")
```

## **Problem 20 (Thinking-Based / Application)**
### **The Security Log Trim (Tail Analysis)**
A cloud security server generates system log entries continuously:
`logs = ["LOG1", "LOG2", "LOG3", "LOG4", "LOG5", "LOG6", "LOG7", "LOG8", "LOG9", "LOG10"]`.
* Use negative slicing to extract the last 5 logs into a new variable `recent_logs`.
* Reverse `recent_logs` using the `.reverse()` method.
* Loop through the reversed list and print each log entry prefixed with `"CRITICAL AUDIT: "` using a `for` loop.

```python
logs = ["LOG1", "LOG2", "LOG3", "LOG4", "LOG5", "LOG6", "LOG7", "LOG8", "LOG9", "LOG10"]

# 1. Negative slice to extract tail (last 5 entries)
recent_logs = logs[-5:]

# 2. Reverse list in-place to get reverse chronological order
recent_logs.reverse()

# 3. Print each log with formatted prefix
for log in recent_logs:
    print(f"CRITICAL AUDIT: {log}")
```
