## **Problem 7 (Moderate)**
Write a program that divides 27 by 5.

Print the exact decimal result using normal division, and the whole number result using floor division (`//`) on the next line.

```python
dividend = 27
divisor = 5

# Normal division (/) preserves the complete precision and returns a float
exact_result = dividend / divisor

# Floor division (//) truncates the decimal components, rounding down to the nearest integer
whole_result = dividend // divisor

# Display the computed behaviors on consecutive lines
print(exact_result)
print(whole_result)
```

## **Problem 11 (Thinking-Based)**
A user wants to withdraw a specific amount of money from an ATM (e.g., rupees 235).

Because the machine only handles 100 rupees bills, it can only dispense amounts that are perfect multiples of 10.

Write a program using arithmetic and relational operators that outputs True if the amount is valid for withdrawal and False if it isn't.

```python
# Accept the withdrawal amount from the user
amount = int(input("Enter the amount to withdraw: "))

# To be valid, the amount must be a multiple of 10 (perfectly divisible with no remainder).
# The modulus operator (%) calculates the remainder. If remainder is 0, it is valid.
is_valid_amount = (amount % 10 == 0)

# Print the final validation outcome
print("Is the withdrawal amount valid?", is_valid_amount)
```

## **Problem 12 (Thinking-Based)**
A bakery recipe requires exactly 3 eggs to make one batch of cookies. The baker looks in the fridge and finds they have `total_eggs = 14`.

Write a program using arithmetic operators to calculate exactly how many full batches they can make, and how many individual eggs will be left over.

```python
total_eggs = 14
eggs_per_batch = 3

# Floor division (//) divides and discards the remainder to give the number of full batches
batches_possible = total_eggs // eggs_per_batch

# Modulus (%) isolates the remainder, showing exactly how many eggs are left over
eggs_remaining = total_eggs % eggs_per_batch

# Display the final breakdown clearly
print("Total full batches possible:", batches_possible)
print("Individual eggs left over:", eggs_remaining)
```

## **Problem 13 (Thinking-Based)**
You are programming a smart home display. The outside temperature sensor reads `temperature = 18` degrees Celsius.

Write an expression combining relational and logical operators to check if the weather is in the "ideal jacket range" (between 15 and 22 degrees inclusive), printing True if it is.

```python
temperature = 18

# We check if the temperature is greater than or equal to 15 AND less than or equal to 22
# The 'and' operator makes sure both bounds are respected simultaneously
is_ideal_range = (temperature >= 15) and (temperature <= 22)

# Alternatively, Python allows direct chaining: is_ideal_range = 15 <= temperature <= 22

# Print the structural check outcome
print("Is the temperature in the ideal jacket range?", is_ideal_range)
```

## **Problem 14 (Thinking-Based)**
To enter the VIP section of a festival, a guest must either have a `vip_pass` (True/False) OR be over the age of 21. However, if they are on the `banned_list` (True/False), they cannot enter under any circumstances.

Create these three variables with sample values and write a single logical expression to determine if access is granted.

```python
# Define state variables with sample data for verification
has_vip_pass = True
age = 19
is_banned = False

# The logic demands either a VIP pass OR age > 21 to clear the first gate.
# Crucially, they must NOT be on the banned list, which applies to everyone.
# We use parentheses to ensure proper execution priority (precedence) for the OR condition.
access_granted = (has_vip_pass or age > 21) and (not is_banned)

# Output the access evaluation
print("Access to VIP Section Granted:", access_granted)
```

## **Problem 15 (Thinking-Based)**
A website registration page requires all new usernames to be secure. The rule states a username must be longer than 5 characters but no longer than 15 characters.

Take a string input for a username, calculate its length, and use operators to output whether the username fits the required security rules.

```python
# Accept the incoming username string input
username = input("Enter your desired username: ")

# The built-in len() function calculates the total character count of the string
username_length = len(username)

# The rule dictates: length > 5 AND length <= 15
is_secure_username = (username_length > 5) and (username_length <= 15)

# Display the validation status to the console
print("Does the username meet security criteria?", is_secure_username)
```
