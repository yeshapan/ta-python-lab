## **Problem 21**

A telemetry system records continuous location coordinates as tuples inside a list: route =
[(18.52, 73.85), (19.07, 72.87), (12.97, 77.59)].
* Write a program that loops through route.
* Unpack each coordinate tuple into latitude and longitude.
* Print a formatted log string: "Checkpoint X -> Latitude: Y, Longitude: Z".
* Calculate the average latitude across all checkpoints using a loop.

```python
route = [
    (18.52, 73.85),
    (19.07, 72.87),
    (12.97, 77.59)
]

total_latitude = 0
count = 0

for coordinate in route:
    latitude, longitude = coordinate
    count += 1
    total_latitude += latitude

    print(f"Checkpoint {count} -> Latitude: {latitude}, Longitude: {longitude}")

average_latitude = total_latitude / count

print("Average Latitude:", average_latitude)
```

## **Problem 22**

An online store tracks items selected by a user in a dictionary where the key is the product
name and the value is its price: cart = {"Shirt": 800, "Jeans": 1500, "Shoes":
2200, "Hat": 300}.
* Ask the user if they have a promo code.
* If the promo code is "DISCOUNT10", apply a 10% discount to all items by updating the
values in the dictionary.
* Calculate the final total bill by looping through the keys of cart and adding up
cart[key].
* If the total bill exceeds 3000, print "Eligible for Free Shipping!", otherwise
print "Standard Shipping Applied".

```python
cart = {
    "Shirt": 800,
    "Jeans": 1500,
    "Shoes": 2200,
    "Hat": 300
}

promo = input("Do you have a promo code? ")

if promo == "DISCOUNT10":
    for item in cart:
        cart[item] = cart[item] * 0.90

total = 0

for item in cart:
    total += cart[item]

print("Final Total Bill:", total)

if total > 3000:
    print("Eligible for Free Shipping!")
else:
    print("Standard Shipping Applied")
```

## **Problem 23**

A bank branch stores customer data using account numbers as keys: accounts = {1001:
{"name": "Priya", "balance": 5000}, 1002: {"name": "Rahul", "balance":
12000}}.
* Ask the user to enter their account number.
* Use the in operator to check if the account exists.
* If it exists, ask whether they want to "deposit" or "withdraw", along with the
amount.
* Update the "balance" key inside the nested dictionary directly using
accounts[acc_num]["balance"].
* If a withdrawal exceeds the available balance, cancel the transaction and print
"Insufficient Funds".

```python
accounts = {
    1001: {"name": "Priya", "balance": 5000},
    1002: {"name": "Rahul", "balance": 12000}
}

acc_num = int(input("Enter account number: "))

if acc_num in accounts:

    choice = input("Enter deposit or withdraw: ")
    amount = float(input("Enter amount: "))

    if choice == "deposit":
        accounts[acc_num]["balance"] += amount
        print("Deposit successful!")
        print("New Balance:", accounts[acc_num]["balance"])

    elif choice == "withdraw":
        if amount <= accounts[acc_num]["balance"]:
            accounts[acc_num]["balance"] -= amount
            print("Withdrawal successful!")
            print("New Balance:", accounts[acc_num]["balance"])
        else:
            print("Insufficient Funds")

    else:
        print("Invalid choice")

else:
    print("Account does not exist")
```

A boutique hotel manages room statuses using a dictionary: rooms = {101: "Occupied",

## **Problem 25**

You are designing the backend for a mobile phonebook app: contacts = {"Amit":
"9876543210", "Neha": "9123456789", "Suresh": "9988776655"}.
* Ask the user for a contact name to search.
* If the name exists in contacts (using in), ask if they want to update the phone number
or delete the contact.
* If they choose update, take a new number input and update the entry using .update().
* If they choose delete, use .pop() to remove the key and print "Contact deleted
successfully."
* If the name does not exist, ask if they want to add it as a new contact.

```python
contacts = {
    "Amit": "9876543210",
    "Neha": "9123456789",
    "Suresh": "9988776655"
}

name = input("Enter contact name: ")

if name in contacts:

    choice = input("Do you want to update or delete? ")

    if choice == "update":
        new_number = input("Enter new phone number: ")

        contacts.update({name: new_number})

        print("Contact updated successfully.")
        print(contacts)

    elif choice == "delete":
        contacts.pop(name)

        print("Contact deleted successfully.")
        print(contacts)

    else:
        print("Invalid choice")

else:
    choice = input("Contact not found. Do you want to add it? ")

    if choice == "yes":
        number = input("Enter phone number: ")
        contacts[name] = number

        print("Contact added successfully.")
        print(contacts)
    else:
        print("Contact not added.")
```

## **Problem 26**

A school library tracks available copies of books: library = {"Python Basics": 4,
"Data Structures": 2, "Web Dev": 0}.
* A student wants to borrow "Web Dev". Check if the book exists in library using in
and if its count library["Web Dev"] is greater than 0.
* If the count is 0, print "Book currently out of stock."
* If a student returns 2 copies of "Python Basics", update the inventory count by
adding 2 to library["Python Basics"].
* Automatically delete any book entry from library using del if its inventory count ever
drops to 0.

```python
library = {
    "Python Basics": 4,
    "Data Structures": 2,
    "Web Dev": 0
}

# Check Web Dev availability
if "Web Dev" in library:
    if library["Web Dev"] > 0:
        print("Web Dev is available.")
    else:
        print("Book currently out of stock.")
else:
    print("Book does not exist.")

# Student returns 2 copies of Python Basics
library["Python Basics"] += 2

# Delete books whose count becomes 0
for book in list(library):
    if library[book] == 0:
        del library[book]

print("Updated Library:", library)
```

## **Problem 27**

An election poll uses voter records where keys are tuples containing (voter_id,
region_code) to prevent duplicate voting: voter_registry = {(101, "WEST"):
"Voted", (102, "EAST"): "Voted"}.
* A new voter arrives with voter_id = 101 and region_code = "WEST".
* Create a tuple current_voter = (101, "WEST").
* Check if current_voter exists in voter_registry using the in operator.
* If it exists, print "Error: Voter has already cast a vote!"
* If it does not exist, add current_voter to voter_registry with value "Voted" and
print "Vote recorded successfully!"

```python
voter_registry = {
    (101, "WEST"): "Voted",
    (102, "EAST"): "Voted"
}

voter_id = 101
region_code = "WEST"

current_voter = (voter_id, region_code)

if current_voter in voter_registry:
    print("Error: Voter has already cast a vote!")
else:
    voter_registry[current_voter] = "Voted"
    print("Vote recorded successfully!")
```

## **Problem 28**

A warehouse assigns task queues to workers using a dictionary ordered by shift priority:
task_queue = {"Task1": "Unload Truck", "Task2": "Inspect Boxes",
"Task3": "Label Packages"}.

* Write a while loop that processes tasks one by one from the end of the queue using
.popitem().
* In each iteration, unpack the popped tuple into task_id and task_desc, then print
"Processing and removing: [task_id] -> [task_desc]".
* Stop the loop when len(task_queue) == 0 and print "All shift tasks
completed!"

```python
task_queue = {
    "Task1": "Unload Truck",
    "Task2": "Inspect Boxes",
    "Task3": "Label Packages"
}

while len(task_queue) > 0:

    task_id, task_desc = task_queue.popitem()

    print("Processing and removing:", task_id, "->", task_desc)

print("All shift tasks completed!")
```

## **Problem 29**

Two academic branches merge their mid-term score sheets: branch1 = {"StudentA": 78,
"StudentB": 85, "StudentC": 90} branch2 = {"StudentB": 88, "StudentD":
92} (Note: StudentB took tests in both branches).
* Create a copy of branch1 named final_gradebook.
* Loop through the keys of branch2. If a student already exists in final_gradebook,
update their score only if the new score in branch2 is higher.
* If the student does not exist in final_gradebook, add them.
* Print the merged final_gradebook.

```python
branch1 = {
    "StudentA": 78,
    "StudentB": 85,
    "StudentC": 90
}

branch2 = {
    "StudentB": 88,
    "StudentD": 92
}

final_gradebook = branch1.copy()

for student in branch2:

    if student in final_gradebook:

        if branch2[student] > final_gradebook[student]:
            final_gradebook[student] = branch2[student]

    else:
        final_gradebook[student] = branch2[student]

print("Final Gradebook:", final_gradebook)
```

## **Problem 30**

A smart refrigerator tracks grocery levels in percentages: pantry = {"Milk": 15,
"Eggs": 80, "Butter": 5, "Cheese": 50, "Juice": 10}.
* Create two empty lists: critical_restock and sufficient_stock.
* Iterate through the items in pantry using a for loop over its keys.
* If pantry[item] is below 20%, add a tuple (item, pantry[item]) to
critical_restock.
* If pantry[item] is 20% or higher, append the item name to sufficient_stock.
* Print both lists clearly formatted for the user's shopping display.

```python
pantry = {
    "Milk": 15,
    "Eggs": 80,
    "Butter": 5,
    "Cheese": 50,
    "Juice": 10
}

critical_restock = []
sufficient_stock = []

for item in pantry:

    if pantry[item] < 20:
        critical_restock.append((item, pantry[item]))
    else:
        sufficient_stock.append(item)

print("Critical Restock:", critical_restock)
print("Sufficient Stock:", sufficient_stock)
```
