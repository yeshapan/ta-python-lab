## **Problem 20**

Write a program using a while loop that continuously takes
numerical inputs from a user and appends them to a list. Stop
when the user enters -1. Then, write a custom function
find_highest(lst) that finds and returns the largest number in the
list without using the built-in max().

```python
def find_highest(lst):
    if not lst:
        return None
    highest = lst[0]
    for num in lst:
        if num > highest:
            highest = num
    return highest

user_numbers = []
while True:
    val = float(input("Enter a number (-1 to stop): "))
    if val == -1:
        break
    user_numbers.append(val)

if user_numbers:
    print("Largest number entered:", find_highest(user_numbers))
```

## **Problem 21**

An online storefront stores product listings as a list of dictionaries:
products = [{"name": "Laptop", "price": 55000}, {"name": "Mouse", "price": 800}, {"name":
"Keyboard", "price": 2500}].
* Write a program that asks the user whether they want to sort by "price" or by "name".
* Write two helper functions, sort_by_price(item) and sort_by_name(item).
* Pass the appropriate function to sorted() using the key parameter based on the user's
choice.
* Print the newly sorted product list clearly.

```python
products = [
    {"name": "Laptop", "price": 55000},
    {"name": "Mouse", "price": 800},
    {"name": "Keyboard", "price": 2500}
]

def sort_by_price(item):
    return item["price"]

def sort_by_name(item):
    return item["name"]

choice = input("Sort by 'price' or 'name': ").strip().lower()

if choice == "price":
    sorted_products = sorted(products, key=sort_by_price)
elif choice == "name":
    sorted_products = sorted(products, key=sort_by_name)
else:
    print("Invalid choice! Showing default list.")
    sorted_products = products

print("Sorted Product List:")
for product in sorted_products:
    print(product)
```

## **Problem 22**

An HR system maintains employee salary tuples inside a dictionary:
payroll = {"E101": (50000, "Tech"), "E102": (75000, "Executive"), "E103": (30000, "Tech")}.
* Write a custom function calculate_tax(salary) that applies a 10% tax rate if salary is
50,000 or less, and 20% if above 50,000.
* Loop through the keys of payroll.
* Unpack each tuple into salary and dept.
* Calculate the net salary using calculate_tax() and update payroll[emp_id] to store a new
tuple: (original_salary, net_salary, dept).

```python
payroll = {
    "E101": (50000, "Tech"),
    "E102": (75000, "Executive"),
    "E103": (30000, "Tech")
}

def calculate_tax(salary):
    if salary <= 50000:
        return salary * 0.10
    else:
        return salary * 0.20

for emp_id in payroll:
    salary, dept = payroll[emp_id]
    tax = calculate_tax(salary)
    net_salary = salary - tax
    payroll[emp_id] = (salary, net_salary, dept)

print("Updated Payroll Ledger:")
print(payroll)
```

## **Problem 24**

A telemetry system receives raw sensor data with corrupted negative readings:
raw_readings = [12.4, -999.0, 15.6, 18.2, -999.0, 22.1, 14.0].
* Write a function clean_and_round_data(data) that:
1. Filters out all negative readings ($< 0$).
2. Rounds all remaining valid readings to the nearest whole integer using round().
* Return and print the final cleaned list of readings.

```python
raw_readings = [12.4, -999.0, 15.6, 18.2, -999.0, 22.1, 14.0]

def clean_and_round_data(data):
    cleaned = []
    for reading in data:
        if reading >= 0:
            cleaned.append(round(reading))
    return cleaned

cleaned_readings = clean_and_round_data(raw_readings)
print("Cleaned Readings:", cleaned_readings)
```

## **Problem 25**

Two distribution centers track stock levels:
warehouse_A = {"Items_A": 150, "Items_B": 80, "Items_C": 40}
warehouse_B = {"Items_B": 50, "Items_C": 60, "Items_D": 200}.
* Create a duplicate copy of warehouse_A named master_stock.
* Loop through the keys of warehouse_B. If an item exists in master_stock, add the stock
quantity from warehouse_B to the existing amount.
* If it doesn't exist, add the new item key-value pair.
* Write a custom key function get_stock(item_key) to sort and print master_stock items
from highest stock to lowest stock using sorted().

```python
warehouse_A = {"Items_A": 150, "Items_B": 80, "Items_C": 40}
warehouse_B = {"Items_B": 50, "Items_C": 60, "Items_D": 200}

master_stock = warehouse_A.copy()

for item in warehouse_B:
    if item in master_stock:
        master_stock[item] += warehouse_B[item]
    else:
        master_stock[item] = warehouse_B[item]

def get_stock(item_key):
    return master_stock[item_key]

sorted_keys = sorted(master_stock, key=get_stock, reverse=True)

print("Master Inventory (Highest to Lowest):")
for item in sorted_keys:
    print(f"{item}: {master_stock[item]} units")
```

## **Problem 27**

A school administration records student performance as a list of dictionaries:
gradebook = [{"name": "Rohan", "score": 82}, {"name": "Pooja", "score": 45}, {"name": "Neha",
"score": 91}, {"name": "Amit", "score": 38}].
* Write a function process_gradebook(records) that:
1. Builds a list retest_list containing student dictionaries for those who scored below
50.
2. Builds a list of formatted strings like "Rohan: PASS" (if score $\ge 50$) or "Pooja:
FAIL" (if score $< 50$).

* Return both lists from the function and print them.

```python
gradebook = [
    {"name": "Rohan", "score": 82},
    {"name": "Pooja", "score": 45},
    {"name": "Neha", "score": 91},
    {"name": "Amit", "score": 38}
]

def process_gradebook(records):
    retest_list = []
    status_list = []

    for student in records:
        if student["score"] < 50:
            retest_list.append(student)
            status_list.append(f"{student['name']}: FAIL")
        else:
            status_list.append(f"{student['name']}: PASS")

    return retest_list, status_list

retest, status = process_gradebook(gradebook)
print("Retest Required List:", retest)
print("Status Summaries:", status)
```

```python
devices = {"Light": "OFF", "AC": "ON", "Fan": "ON", "Heater": "OFF"}

def toggle_device_status(current_status):
    return "OFF" if current_status == "ON" else "ON"

user_choice = input("Action (type 'Toggle All' to invert device states): ").strip()

if user_choice.lower() == "toggle all":
    for device_name in devices:
        devices[device_name] = toggle_device_status(devices[device_name])

print("Final Device Statuses:", devices)
```

## **Problem 29**

A cybersecurity server records incoming request metadata as tuples:
logs = [("192.168.1.1", "GET", 200), ("10.0.0.5", "POST", 500), ("192.168.1.1", "POST", 403),
("172.16.0.2", "GET", 500)].
* Write a function detect_server_errors(log_list) that loops through the logs and extracts all
tuples where the HTTP status code (3rd element) is 500.
* Count how many internal server errors (code 500) occurred.
* Print the isolated error logs along with the total count.

```python
logs = [
    ("192.168.1.1", "GET", 200),
    ("10.0.0.5", "POST", 500),
    ("192.168.1.1", "POST", 403),
    ("172.16.0.2", "GET", 500)
]

def detect_server_errors(log_list):
    error_logs = []
    for log in log_list:
        if log[2] == 500:
            error_logs.append(log)
    return error_logs, len(error_logs)

isolated_errors, count = detect_server_errors(logs)
print(f"Total Internal Server Errors (500): {count}")
print("Isolated Error Logs:", isolated_errors)
```

## **Problem 30**

A digital wallet tracks transaction history as a list of tuples containing transaction type and
amount:
transactions = [("Deposit", 1000), ("Withdrawal", 200), ("Withdrawal", 150), ("Deposit", 500)].
* Write a function process_ledger(transaction_list) that starts with balance = 0.
* Loop through transactions, unpacking each tuple into t_type and amount.
* Add deposits and subtract withdrawals from balance.
* Write a second function is_solvent(balance) that returns True if balance >= 0 and False
otherwise.
* Print the final balance and the result of is_solvent(balance).

```python
transactions = [
    ("Deposit", 1000),
    ("Withdrawal", 200),
    ("Withdrawal", 150),
    ("Deposit", 500)
]

def process_ledger(transaction_list):
    balance = 0
    for t_type, amount in transaction_list:
        if t_type == "Deposit":
            balance += amount
        elif t_type == "Withdrawal":
            balance -= amount
    return balance

def is_solvent(balance):
    return balance >= 0

final_balance = process_ledger(transactions)
solvency_status = is_solvent(final_balance)

print(f"Final Balance: ${final_balance}")
print(f"Wallet Solvent: {solvency_status}")
```
